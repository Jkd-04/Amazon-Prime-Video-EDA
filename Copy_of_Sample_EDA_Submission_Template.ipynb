## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.
import ast

# 1. Selecting important columns and creating a working copy

columns_needed = [
    'id', 'title', 'type', 'release_year', 'genres',
    'production_countries', 'runtime', 'seasons',
    'age_certification', 'imdb_score', 'imdb_votes',
    'tmdb_popularity', 'tmdb_score'
]

df_clean = df2[columns_needed].copy()


# 2. Removing duplicate records

df_clean.drop_duplicates(inplace=True)
df_clean.reset_index(drop=True, inplace=True)


# 3. Checking missing values

missing_analysis = pd.DataFrame({
    'Missing Values': df_clean.isnull().sum(),
    'Missing Percentage': (
        df_clean.isnull().sum() / len(df_clean) * 100
    ).round(2)
})

print("Missing Value Analysis:")
print(missing_analysis)


# 4. Handling missing categorical values

df_clean['age_certification'] = (
    df_clean['age_certification'].fillna('Unknown')
)


# 5. Correcting data types

df_clean['seasons'] = df_clean['seasons'].astype('Int64')
df_clean['imdb_votes'] = df_clean['imdb_votes'].astype('Int64')

df_clean['type'] = df_clean['type'].astype('category')
df_clean['age_certification'] = (
    df_clean['age_certification'].astype('category')
)


# 6. Standardising categorical values

df_clean['type'] = (
    df_clean['type'].str.strip().str.upper()
)

df_clean['age_certification'] = (
    df_clean['age_certification'].str.strip().str.upper()
)


# 7. Converting genres and production countries into lists

def parse_list(value):
    if isinstance(value, list):
        return value
    if pd.isna(value):
        return []
    return ast.literal_eval(value)

df_clean['genres'] = df_clean['genres'].apply(parse_list)

df_clean['production_countries'] = (
    df_clean['production_countries'].apply(parse_list)
)


# 8. Checking skewness in numerical columns

numeric_cols = [
    'runtime', 'imdb_score', 'imdb_votes',
    'tmdb_popularity', 'tmdb_score'
]

skewness = df_clean[numeric_cols].skew()

print("\nSkewness Analysis:")
print(skewness)


# 9. Identifying potential outliers using IQR

outlier_summary = []

for col in numeric_cols:

    Q1 = df_clean[col].quantile(0.25)
    Q3 = df_clean[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outlier_count = (
        (df_clean[col] < lower_limit) |
        (df_clean[col] > upper_limit)
    ).sum()

    outlier_summary.append({
        'Column': col,
        'Outlier Count': outlier_count
    })

outlier_summary = pd.DataFrame(outlier_summary)

print("\nPotential Outlier Analysis:")
print(outlier_summary)


# 10. Final dataset verification

print("\nFinal Dataset Shape:", df_clean.shape)

# Exclude unhashable list columns (genres, production_countries) when checking duplicates
hashable_cols = [col for col in df_clean.columns if col not in ['genres', 'production_countries']]
print("Duplicate Records (excluding list columns):", df_clean.duplicated(subset=hashable_cols).sum())

print("\nFinal Missing Values:")
print(df_clean.isnull().sum())

print("\nData Types:")
print(df_clean.dtypes)

print("\nData Wrangling Completed Successfully!")
````

### What all manipulations have you done and insights you found?


The Amazon Prime Video titles dataset underwent data preparation to improve consistency and ensure its suitability for exploratory analysis. The process involved selecting relevant variables, creating a working copy, removing duplicate records, examining missing values, correcting data types and standardising categorical information. Missing age certifications were replaced with 'Unknown', while unavailable numerical metrics were retained to avoid introducing artificial values.

The genres and production countries columns were converted into Python lists to facilitate further analysis. Skewness and potential outliers were also examined using statistical methods, without automatically removing unusual observations.

Initial analysis revealed that the dataset contained 9,871 records, including 8,514 movies and 1,357 TV shows. Three exact duplicate records were identified. The release years ranged from 1912 to 2022, and the average runtime was approximately 86 minutes. Missing season values were observed exclusively for movies, indicating that this information was not applicable to those records. Both IMDb and TMDB scores had average values of approximately 5.98.

````python
# Correlation Heatmap visualization code

# Chart 14: Correlation Heatmap of Numerical Variables

# Selecting important numerical columns
numeric_cols = [
    'release_year',
    'runtime',
    'seasons',
    'imdb_score',
    'imdb_votes',
    'tmdb_popularity',
    'tmdb_score'
]

# Calculating the correlation matrix
correlation_matrix = df_clean[numeric_cols].corr(method='pearson')

# Creating the correlation heatmap
plt.figure(figsize=(11, 8))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5,
    vmin=-1,
    vmax=1,
    center=0
)

plt.title('Correlation Heatmap of Amazon Prime Video Titles')
plt.tight_layout()
plt.show()
````

#### Chart - 15 - Pair Plot

````python
# Pair Plot visualization code
# Chart 15: Pair Plot of Numerical Variables

# Selecting important numerical variables
pairplot_data = df_clean[
    ['runtime', 'imdb_score', 'imdb_votes',
     'tmdb_popularity', 'tmdb_score']
].dropna().copy()

# Applying log transformation to highly skewed variables
pairplot_data['imdb_votes'] = np.log1p(pairplot_data['imdb_votes'])

pairplot_data['tmdb_popularity'] = np.log1p(
    pairplot_data['tmdb_popularity']
)

# Creating the pair plot
sns.pairplot(
    pairplot_data,
    diag_kind='hist',
    corner=True,
    plot_kws={'alpha': 0.4, 's': 18}
)

plt.suptitle(
    'Pair Plot of Numerical Variables in Amazon Prime Video Dataset',
    y=1.02
)

plt.show()

````

##### 1. Why did you pick the specific chart?

A Pair Plot was selected to examine the individual distributions of important numerical variables and understand the relationships between them. It provides a consolidated visual representation of patterns, correlations, data concentration and potential outliers across multiple variables, helping identify relationships that may require further investigation.

##### 2. What is/are the insight(s) found from the chart?

- IMDb Score and TMDB Score: A positive relationship is visible, suggesting that titles with higher IMDb ratings generally tend to receive higher TMDB ratings.
- IMDb Votes and TMDB Popularity: A positive association is visible, indicating that titles with more IMDb votes tend to show higher TMDB popularity.
- IMDb Votes Distribution: The distribution is right-skewed, with most titles receiving comparatively fewer votes and a small number receiving very high vote counts.
- TMDB Popularity Distribution: Popularity is also right-skewed, with a few titles showing exceptionally high popularity.
- Runtime and Ratings: No strong linear relationship is visually apparent between runtime and IMDb or TMDB scores.
Overall finding: The chart suggests that IMDb and TMDB ratings, as well as IMDb votes and TMDB popularity, show positive associations.
